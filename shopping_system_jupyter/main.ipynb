```python
store = {
    'flour': {'price': 140, 'quantity': 1000}, 
    'rice': {'price': 400, 'quantity': 150}, 
    'oil': {'price': 450, 'quantity': 150}, 
    'milk': {'price': 240, 'quantity': 250}, 
    'sugar': {'price': 250, 'quantity': 150} 
}

item = input("What do you want Sir ? : ").lower()

for index in store:
    if index == item:
        print(f"\n{item.title()} Available!")
        
        print(f"{store[item]['price']}Rs per KG")
        
        quantity = eval(input("How many kgs Sir ? : "))
        
        if quantity <= store[item]['quantity']:
            
            bill = quantity * store[item]['price']
            
            
            confirm=input(f"Do you want to confirm your order of {item.title()} {quantity}kgs at  Rs {bill} \n (Y/N)").lower()
            
            if confirm =="y":
                store[item]['quantity'] -= quantity
                print(f"Order Confirmed !")
                print(f"Please pay bill: Rs {bill}")
            else:
             print(f"Order Cancelled!")  
             
            
        else:
            print(f"The {item.title()} at {quantity}kg is not available, sir.")
            
        break  
else:
    print(f"Sir, {item} is not available.")
```
