## Data Frame


```python
import pandas as pd

data={
    "Name": ["Umair","Ayaan","Zain"],
    "Age": [17,20,16],
    "Marks":[100,100,45]
}

df=pd.DataFrame(data,index=["Student 1","Student 2","Student 3"])

# Add A Column

df["Job"]=["Data Scientist","AI Engineer","Aerospace Engineer"]

# Add A Row

# new_row = pd.DataFrame([{"Name":"Bacha","Age":12,"Marks":20,"Job":"Berozgar"}],index=["Student 4"])

# df=pd.concat([df,new_row])


# Add Rows

new_rows = pd.DataFrame([{"Name":"Bacha","Age":12,"Marks":20,"Job":"Berozgar"},{"Name":"Chacha","Age":52,"Marks":10,"Job":"Fire Fighter"}],index=["Student 5","Student 6"])

df=pd.concat([df,new_rows])
print(df)
```
