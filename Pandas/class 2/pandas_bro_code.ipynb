```python
import pandas as pd

data=[10,23,34]

series=pd.Series(data,index=["a","b","c"])

# loc - means location , by label it is used to call the value at the index code below if not present keyerror.
series.loc["b"] # acces loc by this 

series.loc["b"]=23 # the value can be changed also 

# iloc - means integer location code to below
# can be called like this
print(series.iloc[0])
```

## Filter Data

```python
import pandas as pd

data=[100,150,120,200,340,200]

series=pd.Series(data,index=["a","b","c","d","e","f"])

# so to call values that are greater than 200

# print(series[series>=200])


# so to call values that are less than 200

print(series[series <=200])
```

## Filter Dictionary

```python
import pandas as pd

calories={"Day1":2000,
          "Day2":1500,
          "Day3":1000}

series= pd.Series(calories) # index ni dalna q k key index ka kaam kr rhi hai

# loc in dictionary

# print(series.loc["Day1"])

# also can be used by doing this print(series["Day1"])
# to add data on value we will do this

series.loc["Day2"]+= 1500

# filter like how many days we have eaten more than 2000
print(series[series >=2000])
```
