## Data Frame
- data ma aik sai zyada column ho to date frame bn jata hai

```python
data = {
    'Name':['Umair','Fasih Ullah','Aizaz','Owais','Zain','Faizan'],
    'Roll_Number':[12,23,34,56,12,23],
    'Python':[0,2,3,4,5,6]
}
data=pd.DataFrame(data)
```

```python
data
```
