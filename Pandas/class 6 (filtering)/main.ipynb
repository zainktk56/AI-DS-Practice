## FILTERING DATA PANDAS



#### FILTERING - MEANS KEPP THE ROWS THAT MATCHES THE CONDITION

```python
import pandas as pd

df= pd.read_csv("data.csv")

# this prints the height of the pokemon that prints the height of the pokemon that are greater than 2 meters
print(df[df["Height"] >= 2])


# to filter weight

print(df[df["Weight"] <= 100 ])




```
