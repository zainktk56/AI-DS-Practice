```python

class Computer:
    def process(self):
        pass # a method that only has declaration but has nothing in it is method , A class having method that has no body
```

```python
comp1= Computer()
comp1.process()

# there is no error and we are able to create object and call method , because it is not an abstract class and not an abstract method.
```

```python
# to do abstraction we will use ABC module

from abc import ABC, abstractmethod

class Computer(ABC):
    @abstractmethod
    def process(self):
        pass
    @abstractmethod
    def greet(self): # it is also a abstract class because it have an abstract method
        print("hello")
        
        # to make class abstract
        # - it must inherit the ABC class from abc module
        # - it must have atleast a abstract method
        # - which is defined as a decorator (@abstractmethod)
        # - it does not have object
        # - it does not have a body
       
        
```

```python
com1=Computer()
```
