```python
with open('Friends_Name.txt','a') as f:
    counter=1
    while True:
     name=input('Enter a name or (q to quit): ')
     if name=='q':
         break
     else:
          f.write(f"  {counter}-  {name} \n")
          counter+=1
          
```
