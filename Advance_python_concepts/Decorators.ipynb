# Decorators


### Python decorators allow you to modify or extend the behaviour of funtions and method without changing its code  .

```python
def say_hello(name):
    print (f"Hello ,{name}")
def greet(name):
    print(f"YO how are you,{name}")
    
def greet_bob(func):
    return func("bob")        
```

```python
greet_bob(len)
```

```python
# this is a decorator function . it accepts a function as a parameter
def my_decorator(func):
    def wrapper(): # a inner function . usually called as a wrapper 
        print("Something is happening before the function is called.")
        func()
        print("Something is happening after the function is called.")
    return wrapper

@my_decorator
def greet():
    print("Hello, World!")

greet()
```
