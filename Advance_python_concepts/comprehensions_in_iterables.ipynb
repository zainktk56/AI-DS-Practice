# List Comprehension


###### - Provides a concise one line syntax to create a new list from an existing iterable

```python
list=[1,23,4,4,5,6,6,7,78,86,5,5]

sq_list=[]

for n in list:
    sq= n*n
    sq_list.append(sq)
sq_list    
```

```python
## Same thing can be done in one line 

[n**2 for n in list]
```

```python
list=[1,23,4,4,5,6,6,7,78,86,5,5]

new_list=[]
for n in list:
    if n%2==0:
        sq=n*n
        new_list.append(sq)
    else:
        cub=n*n*n
        new_list.append(cub)   
        
new_list         
```

```python
[ n*n if n%2==0 else  n**3   for n in list]
```

```python
percentage=[45,57,67,89,34,89,56,78]

["A+" if p>=90 else "A" if p>=80 else "B" if p>=70 else "C" if p>=60 else "D" if p>=50 else "F" for p in percentage]
```

# Dictionary Comprehension


```python
{n:n**2 for n in list}
```

```python
country=['pakistan','russia','japan','iran','palestine']
{c:len(c) for c in country}
```
