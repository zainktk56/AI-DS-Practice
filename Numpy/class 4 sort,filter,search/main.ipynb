# Sort,Filter , Search

- Sort Function

```python
import numpy as np

arr=np.array([[56,90,1,45,100],[23,54,1,21,87]])
print(np.sort(arr))
```

- Search Function

```python
ar=np.array([56,90,1,45,100])
# s= np.where(ar == 56)
s= np.where(ar % 2 == 0)
print(s)
```

- Search Sorted Function

```python
arrr=np.array([1,2,4,5,6])
ss=np.searchsorted(arrr,6)
print(ss)
```
