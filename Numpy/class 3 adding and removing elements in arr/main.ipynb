## Adding and Removing Elements In Array

#### Functions
- np.append(h,g) => Append Items to an Array
- np.insert(a,1,5) => Insert Items to a specific position in an array
- np.delete(a,[1]) => Delete Items from an array 

```python
import numpy as np
a=np.array([[10,20],[30,40]])


print(np.append(a,40))
 # the output is coming because numpy first convert this into a list then append it
```

```python
print(a)
print(np.insert(a,1,[40,60],axis=1))# array,index,value
```

```python
print(np.delete(a,1,axis=1))
```
