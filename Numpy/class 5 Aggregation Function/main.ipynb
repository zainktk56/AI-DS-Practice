# Aggregation Function Numpy


```python
import numpy as np

a = np.array([1,2,3,4,5,76])

print(np.sum(a))

print(np.min(a))

print(np.max(a))

print(np.cumsum(a))

print(np.cumprod(a))

print(np.mean(a))

print(np.size(a))
```

```python
import numpy as np

a = np.array([120,340,150,199,300])
b = np.array([2,30,3,22,10])

total_price=np.cumprod([a,b],axis=0)
print(total_price[1].sum()) # totalprice
```
