## Numpy - Numerical Python


### Basic Functions of Numpy
- np.shape() => tells us how many rows and cloumn are there in an array
- np.size()  => tells us the total number of elements present in the array
- np.ndim()  => tells us the dimensions of the array
- len(a)     => tells us the length
- np.dtype() => tells us the data type
- np.astype() => Convert an array into different data type

## How to Create an ARRAY ? 
- Normal list looks like the list=[1,3,54,76]
- Array looks like this arr= [2 4 5 6]
-  this is because the list gets stored in the non-contigeous memory location unlike array gets stored in the contigeous memory locations

```python
import numpy as np

a=[[10,34,45,56],[23,34,45.9,67]]
arr=np.array(a)
print(arr)
print(arr.shape)
print(arr.size)
print(arr.dtype)
print(type(arr))
print(arr.ndim)
print(len(arr))
print(arr.astype(float))
print(arr.astype(str))
print(arr.astype(int))
```

- Remember String cannot be converted into INTEGER

## Mathematical Operation IN NUMPY

```python
import numpy as np

arr1=np.array([10,20,30,40])
arr2=np.array([50,60,70,80])

print(arr1+arr2)

print(np.add(arr1,arr2))
```

```python
print(arr1-arr2)
print(np.subtract(arr1,arr2))

print(arr1*arr2)
print(np.multiply(arr1,arr2))

print(arr1/arr2)
print(np.divide(arr1,arr2))
```

- Power Funcion

```python
arr1=np.array([10,20,30,40])
arr2=np.array([5])
print(np.power(arr1,arr2))
```

- SQAURE ROOT

```python
arr1=np.array([10,20,30,40])

print(np.sqrt(arr1).astype(int))
```
