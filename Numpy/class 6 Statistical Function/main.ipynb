# Statistical Function

```python
import numpy as np
import statistics as stats

baked_price=[100,120,150,200,250]

a = np.array(baked_price)

print(np.mean(a)) # sum of val/ num of val
print(np.median(a)) # it is the middle value
print(stats.mode(a)) # recurring val
print(np.std(a)) # spread between data
print(np.var(a)) # sqaure of standard deviation
```

- Correlation Funtions

```python
# -1 represents inversely proportional
# 1 represents directly proportional
# 0 represents no relation

tobaco_consumption= [30,50,40,10,20]
deaths=[100,120,115,70,90]

print(np.corrcoef([tobaco_consumption,deaths]))
```

```python
# Another Example of the price and sales

price=[100,300,900,200,150,700,500]
sales=[15,5,3,7,9,4,4]
print(np.corrcoef([price,sales]))
```
