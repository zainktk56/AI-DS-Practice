## Joining And Spliting Arrays

#### First of all as we know that when we add two lists they concate but in the numpy we cannot concate them with (+) symbol like in list so to concate there is a reserved function for concatinataion in Numpy

```python
import numpy as np 

arr1=np.array([10,20,30])
arr2=np.array([40,50,60])

print(np.concatenate([arr1,arr2]))
```

# Axis 

- Axis 0  => Means Vertical Axis
- Axis 1  => Means Horizontal Axis

```python
import numpy as np

arr1=np.array([[10,34,23],[11,35,24]])
arr2=np.array([[20,30,50],[60,70,80]])
print(np.concatenate([arr1,arr2],axis=0))

print(np.concatenate([arr1,arr2],axis=1))
```

- Other Method for Horizontal and Vertical Concatenation

```python
# for horizontal concate
print(np.hstack([arr1,arr2]))
```

```python
# for vertical concate
print(np.vstack([arr1,arr2]))
```

```python
a=np.array([10,30,50,30,70])
b=np.array_split(a,2)
print(b)
```
