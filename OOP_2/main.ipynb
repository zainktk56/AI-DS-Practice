# Inheritance


```python
# INHERITANCE

class Car():
    def __init__(self,make,model,year):
        self.make=make
        self.model=model
        self.year=year
        self.odometer_reading=0
        
    def  get_descriptive_name(self):
        long_name=f" {self.year} {self.make} {self.model}"
        return long_name.title()
    
    def read_odometer(self):
        print(f"This Car has {self.odometer_reading} km on it")
        
    def update_odometer(self,milage):
        if milage>self.odometer_reading:
            self.odometer_reading=milage
        else:
            print('YOU CANNOT ROLL BACK ODOMETER')
     
    def increment_odometer(self,increase_milage):
                  self.odometer_reading+=increase_milage
                  print(f"This Car has {self.odometer_reading} km Increased on it")
             
    
                      
```

```python
class ElectricCar(Car):
    def __init__(self, make, model, year, color, hp):
        # Pass parent parameters to the superclass initializer
        super().__init__(make, model, year)
        self.color = color
        self.hp = hp
        self.battery = 100
```

```python
my_e_car = ElectricCar("byd", "atto3", 2009, "blue", 76)
print(my_e_car)
```

```python
my_car=Car('toyota','grande',2000)
print(my_car.increment_odometer(300))
```

```python
class User:

    def __init__(self, first_name, last_name, email, phone_number):
        self.first_name = first_name
        self.last_name = last_name
        self.email = email
        self.phone_number = phone_number
        self.login_attempts = 0  # Plural attribute name avoids collision

    def describe_user(self):
        return f"""
    First Name : {self.first_name}
    Last Name : {self.last_name}
    Email : {self.email}
    Phone Number : {self.phone_number}
    """

    def greet_user(self):
        return f"Welcome {self.first_name}! To Our Learning Portal"

    def increment_login_attempts(self):
        self.login_attempts += 1
        return f"Your login attempts are {self.login_attempts}"

    def reset_login_attempts(self):
        self.login_attempts = 0
        return f"Login attempts reset to {self.login_attempts}"


```

```python
user2 = User("zain", "haroon", "abc@gmail.com", 123)

print(user2.increment_login_attempts())
```
