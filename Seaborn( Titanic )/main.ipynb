# Seaborn Titanic Data Set


```python
import seaborn as sns
import matplotlib.pyplot as plt

data = sns.load_dataset('titanic')
print(data)
```

```python
data.rename({"sex":"gender"},axis=1,inplace=True)
print(data)
```

```python
data=data.drop(["who","embarked","deck","adult_male","pclass"],axis=1)
```

```python
data.isna().sum()
```

```python
data.duplicated().sum()
```

```python
data.drop_duplicates(inplace=True)
```

```python

data.duplicated().sum()
```

```python

data.isna().sum()
```

```python
data.embark_town.value_counts()
```

mean --> outlier is far then filling data medain
median -->  outlier is near than fillinh data mean
mode -- > catagorical values

```python
data=data.fillna({"embark_town":data.embark_town.mode()[0]})
print(data)
```

```python
data.embark_town.isna().sum()
```

```python

sns.boxplot(data.age)
```

```python
data=data.fillna({"age": data.age.mean()})
```

```python
data.age.isna().sum()
```

```python
data
```

```python
x= data.gender.value_counts().index
y= data.gender.value_counts().values
```

```python
plt.title("Gender Distriution",fontdict={"size":30,"color":"g"})
sns.barplot(x=x,y=y,color="g")
plt.xlabel("Gender Data",fontdict={"size":15,"color":"b"})
plt.ylabel("Counts",fontdict={"size":15,"color":"b"})
plt.show()
```

##### Pie Chart

```python
plt.title("Pie Chart (Embark Town) ",fontdict={"size":18,"color":"b"})
plt.pie(x=data.embark_town.value_counts(),labels=data.embark_town.value_counts().index,autopct="%.2f%%",colors=["g","r","b"])
plt.xlabel("Town",fontdict={"size":15,"color":"b"})
plt.ylabel("Frequency",fontdict={"size":15,"color":"b"})
plt.show()
```
