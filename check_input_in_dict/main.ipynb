```python
items={'laptop':50000,'mobile':6000,'earbuds':1500,'headphones':2000,'monitor':10000}

item_name = input("Enter the item name you want to buy: ").lower()

bill_amt=items.get(item_name)

if item_name in items:

    print(f'The Price of {item_name.capitalize()} is {bill_amt}')
    user_input=input('Do you want to buy this item (Y/N) : ').lower()
    if user_input =="y":

        if bill_amt>=20000:
            discount = bill_amt * 0.20
            final_bill = bill_amt - discount
            print(f'Your bill after 20% of discount is {final_bill}')
            print("Thanks for shopping!")

        elif bill_amt>=15000 :
            discount = bill_amt * 0.10
            final_bill = bill_amt - discount
            print(f'Your bill after 10% of discount is {final_bill}')
            print("Thanks for shopping!")

        else:
            discount=bill_amt
            print(f'Your total bill amount is {discount}')
            print("Thanks for shopping!")

    elif user_input=="n":
        print(f'Thank You Goodbye!\n See you next time.')

    else:
        print("invalid input please try again !")

else:
   print(f'{item_name.capitalize()} is not available in our store.')
   
   
       
       









```
